# 지도 데이터 시각화 - 폴리엄

```python
# !pip install folium
import folium
import pandas as pd

```

```python

# 서울시청: [위도, 경도]
city_hall = [37.5662952, 126.9779451]

# 배경 지도 제공자 지정
m = folium.Map(
    location=city_hall,
    zoom_start=14,
    tiles="Esri.WorldImagery"  #  항공,위성영상 사용
    # 배경 지도는 외부 서비스... 영구적인 사용 보장 안됨
)

# 서울시청 마커
folium.Marker(
    location=city_hall,
    tooltip="서울시청",
    popup="서울시청 위치",
    icon=folium.Icon(color="red")
).add_to(m)

m
```

```python
### 스타벅스 매장 위치 표시
url = (
    "https://gist.githubusercontent.com/sarah37/"
    "5504dcb30d717c20678375e1e40dff33/raw/starbucks_unique.csv"
)

df = pd.read_csv(url)

# 필요한 열 확인
display(df[["Store.Name", "Country", "Latitude", "Longitude"]].head())


# Store.Name	매장 이름
# Country	국가 코드 — 한국은 KR
# Latitude	위도
# Longitude	경도
```

```python
# 서울시청 주변 영역의 매장 선택
#  위도,경도의 사각형 범위로 선택 cf, 거리 계산

stores = df[
    (df["Country"] == "KR")  # 국가가 한국인 매장
    & df["Latitude"].between(37.55, 37.58)      # 위도 37.55 이상, 37.58 이하
    & df["Longitude"].between(126.96, 126.99)   # 경도 126.96 이상, 126.99 이하
][["Store.Name", "Latitude", "Longitude"]].copy()

print("선택한 매장 수:", len(stores))
display(stores)

### 아래 열 선택
# Store.Name: 매장 이름
# Latitude: 위도
# Longitude: 경도
```

```python
from folium.plugins import MarkerCluster

cluster = MarkerCluster(name="Starbucks").add_to(m)

# 한 행씩 읽어 매장 위치에 마커 추가
# 첫 번째 매장 읽기 -> 마커 생성, 추가 ... 두 번째 매장 읽기 ->  마커 생성, 추가 .. 모든 매장에 반복
for _, row in stores.iterrows(): #  = for index, row in stores.iterrows():
    # _(언더바) 변수 : 행 인덱스, 반복에는 사용 but 루프 안에서 그 숫자를 직접 가져다 쓰지는 않음 <-- 관례적인 표현
    # 해당 행의 데이터. pandas Series 객체
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        tooltip=row["Store.Name"], # 마우스를 올렸을 때 매장명 표시
        popup=folium.Popup(row["Store.Name"], max_width=250), # 마커를 클릭했을 때 설명 창 표시
                                              # 팝업의 최대 너비를 250픽셀로 설정
        icon=folium.Icon(color="green")
    ).add_to(cluster) # 마커를 클러스터에 추가

display(m)

#### location=[row["Latitude"], row["Longitude"]]
#         데이터의 위도 열      데이터의 경도 열

### 다른 위치 데이터 사용시 ... 매장명, 위도, 경도 열 맞추어 활용
```
